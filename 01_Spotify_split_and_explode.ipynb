```python
df_processed = df.copy()
print(f"Original shape: {df_processed.shape}")

df_processed['music_Influencial_mood'] = df_processed['music_Influencial_mood'].str.split(',')
df_processed = df_processed.explode('music_Influencial_mood')
df_processed['music_Influencial_mood'] = df_processed['music_Influencial_mood'].str.strip()

print(f"Shape after mood explode: {df_processed.shape}")

df_processed['music_lis_frequency'] = df_processed['music_lis_frequency'].str.split(',')
df_processed = df_processed.explode('music_lis_frequency')
df_processed['music_lis_frequency'] = df_processed['music_lis_frequency'].str.strip()

print(f"Shape after frequency explode: {df_processed.shape}")

output_file = 'spotify_data_modified.xlsx'

df_processed.to_excel(output_file, index=False)

df_processed.info()
```

```python
!pip install pandas openpyxl
import pandas as pd
import numpy as np

file_to_load = 'spotify_data_modified_final.xlsx'

df_final = pd.read_excel(file_to_load)
print(f"Successfully loaded. Shape: {df_final.shape}")

print("\n--- Starting: Splitting by 'spotify_subscription_plan' ---")

free_file_name = 'spotify_data_free_users.csv'
premium_file_name = 'spotify_data_premium_users.csv'

df_free = df_final[df_final['spotify_subscription_plan'] == 'Free (ad-supported)']
print(f"\nFound {df_free.shape[0]} rows for 'Free (ad-supported)'.")

df_free.to_csv(free_file_name, index=False)
print(f"Saved to: {free_file_name}")

df_premium = df_final[df_final['spotify_subscription_plan'] == 'Premium (paid subscription)']
print(f"\nFound {df_premium.shape[0]} rows for 'Premium (paid subscription)'.")

df_premium.to_csv(premium_file_name, index=False)
print(f"Saved to: {premium_file_name}")

print("\n--- Finished: Files are split and saved. ---")
```
